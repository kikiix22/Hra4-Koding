```python
# Cvičení: Seznam oblíbených jídel. Uživatel zadá číslo a vypíše se jídlo
# na té pozici. Když číslo v seznamu není, spadne to na IndexError.
# Když projde, vypiš jídlo až v else.


jidla = ["pizza", "špagety", "řízek", "palačinky"]

try:
    cislo = int(input("Zadej číslo jídla (1 až 4): "))
    jidlo = jidla[cislo]
except ValueError:
    print("To nebylo celé číslo.")
except IndexError:
    print("Takové jídlo v seznamu není.")
else:
    print(f"Na pozici {cislo} je {jidlo}.")
    
```

```python
# Projekt
# Textová adventura / „vyber si cestu". Stojíš u jeskyně, 1 vejít, 2 jít dál.
# Každá volba vede jinam, na konci výhra/prohra. Čisté if/elif, while, funkce (každá místnost = funkce).
# Nejvíc kreativní, protože příběh si vymyslí sami, a přitom je to programátorsky pořád jednoduché.
# Jedna rada, kterou bych dal dětem dřív, než napíšou jediný řádek: ať si na papír nakreslí mapu.
# Kolečko = místnost, šipka s číslem = volba. Bez mapy se v tom po třech místnostech ztratí.
# Ta mapa je vlastně přesně ten program, jen nakreslený, a je to hezká ukázka, že „nejdřív návrh, pak kód".

print("Stojíš před temnou jeskyní. Z otvoru táhne chlad.")
print("1) Vejít dovnitř")
print("2) Jít dál po cestě")

volba = input("Co uděláš? ")

if volba == "1":
    print("Vešel jsi do jeskyně a našel truhlu se zlatem. Vyhrál jsi! 🏆")
elif volba == "2":
    print("Šel jsi dál a spadl do skryté jámy. Prohrál jsi.")
else:
    print("To nebyla platná volba. Zaváhal jsi a přišla noc. Konec.")

# Úroveň 2: každá místnost = funkce. Teď se to zlomí do funkcí.
# Místnost něco popíše, zeptá se a podle volby zavolá další místnost. Tím se dá větvit do šířky.

def start():
    print("\nStojíš u vchodu do staré jeskyně.")
    print("1) Vejít  2) Jít dál po cestě")
    volba = input("Co uděláš? ")
    if volba == "1":
        jeskyne()
    elif volba == "2":
        les()
    else:
        print("Zaváhal jsi a je konec.")

def jeskyne():
    print("\nUvnitř jsou dvě chodby.")
    print("1) Levá  2) Pravá")
    volba = input("Kam půjdeš? ")
    if volba == "1":
        print("\nNašel jsi truhlu plnou zlata. Vyhrál jsi! 🏆")
    else:
        print("\nProbudil jsi spícího draka. Konec dobrodružství. 🐉")

def les():
    print("\nCesta končí u hlubokého lesa.")
    print("1) Vejít do lesa  2) Vrátit se")
    volba = input("Co uděláš? ")
    if volba == "1":
        print("\nV lese jsi zabloudil a padla noc. Prohrál jsi.")
    else:
        start()   # vrátíš se na začátek

# spuštění hry
print("=== JESKYNĚ TAJEMSTVÍ ===")
start()
```

```python
# Úroveň 3: while, aby překlep hru neukončil. Zatím každá špatná klávesa rovnou znamená konec.
# To je frustrující. Uděláme pomocnou funkci, která se ptá tak dlouho, dokud nepřijde platná volba.
# To je přesně vzor while True + return z lekce o výjimkách.
def zeptej_se(otazka, platne):
    # ptá se dokola, dokud uživatel nezadá něco ze seznamu platných voleb
    while True:
        volba = input(otazka).strip()   # .strip() ořízne omylem zadané mezery
        if volba in platne:
            return volba
        print("To není platná volba, zkus to znovu.")
```

```python
#Bonus pro rychlíky: inventář. Kdo posbírá věci, otevře si tím nové cesty.
# Ukazuje seznam jako batoh, append (sebrat věc) a in (mám ji?), které rozhodne o pozdějším konci.

inventar = []   # batoh, na začátku prázdný

def sklepeni():
    print("\nNa vlhké zemi leží rezavý klíč.")
    volba = zeptej_se("1) Vzít klíč  2) Nechat být: ", ["1", "2"])
    if volba == "1":
        inventar.append("klíč")
        print("Sebral jsi klíč. 🔑")
    zamcene_dvere()

def zamcene_dvere():
    print("\nStojíš před těžkými zamčenými dveřmi.")
    if "klíč" in inventar:
        print("Odemkl jsi je klíčem a za nimi září poklad. Vyhrál jsi! 🏆")
    else:
        print("Dveře jsou zamčené a ty nemáš čím odemknout. Konec cesty.")

sklepeni()
```
